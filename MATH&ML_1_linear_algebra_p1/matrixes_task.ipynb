```python
import pandas as pd
import numpy as np
```

```python
# Данные по доходам и расходам каждого члена семьи
Husband_Income = np.array([100,220,140])
Wife_Income = np.array([150,200,130])
Mother_In_Law_Income = np.array([90,80,100])

Husband_Consumption = np.array([50,50,60])
Wife_Consumption = np.array([100,80,140])
Mother_In_Law_Consumption = np.array([100,20,140])
```

```python
# матрица с доходами
inc = np.array([Husband_Income, Wife_Income, Mother_In_Law_Income]).T
print(inc)
```

```python
# матрица с расходами
cons = np.array([Husband_Consumption, Wife_Consumption, Mother_In_Law_Consumption]).T
print(cons)
```

```python
# матрица с доходами после вычета налогов
after_tx = inc*0.87
print(after_tx)
```

```python
# матрица с балансом
P = after_tx-cons
print(P)
```

В салоне красоты работают восемь стилистов. Есть две таблицы (DataFrame): Count_DF и Price_DF.
В DataFrame Count_DF содержится информация по количеству услуг, оказанных каждым стилистом за апрель.
В DataFrame Price_DF содержится информация по стоимости услуг у каждого стилиста в тысячах рублей.

```python
Count_DF = pd.DataFrame({
    'Женские стрижки': [10, 2, 12, 4, 6, 10, 22, 7], 
    'Мужские стрижки': [5, 21, 12, 8, 25, 3, 1, 0], 
    'Окрашивания':[12, 3, 0, 18, 27, 2, 4, 31],
    'Укладка':[15, 25, 30, 14, 25, 17, 25, 31],
    'Уход':[10, 6, 4, 5, 18, 12, 20, 28]
    }, 
    index=['Аня', 'Борис', 'Вика', 'Галя', 'Дима', 'Егор', 'Женя','Юра']
)
Price_DF = pd.DataFrame({
    'Женские стрижки': [2, 1.8, 2, 1.8, 2.5, 5, 1.1, 4.5], 
    'Мужские стрижки': [1.5, 2.5, 2, 1.2, 3.5, 5, 1, 4], 
    'Окрашивания':[1, 1, 0, 2.8, 2, 3, 1.5, 2.5],
    'Укладка':[0.8, 1, 0.5, 0.8, 1, 2, 0.5, 1],
    'Уход':[1, 1, 2, 2, 1.5, 2.5, 1.7, 2] 
    }, 
    index=['Аня', 'Борис', 'Вика', 'Галя', 'Дима', 'Егор', 'Женя','Юра']
)

# Вектор комиссий:
com = np.array([0.2, 0.2, 0.3, 0.1, 0.1])
```

```python
Count_DF_values = Count_DF.values
print('количество услуг:\n', Count_DF_values)
```

```python
Price_DF_values = Price_DF.values
print('стоимость услуг:\n', Price_DF_values)
```

```python
Amount_DF = Count_DF_values*Price_DF_values
print('доход от услуг:\n', Amount_DF)
```

```python
Amount_com_DF = Amount_DF@com
print('доход салона:\n', Amount_com_DF)
```

```python
mast = np.ones((1,5))-com
Amount_mast_DF = Amount_DF@mast.T
print('доход мастеров:\n', Amount_mast_DF)
```
