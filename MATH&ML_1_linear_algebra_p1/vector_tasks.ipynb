# Задача 2

```python
# дано три вектора
u = np.array([3,0,1,1,1])
v = np.array([0,1,0,2,-2])
w = np.array([1,-4,-1,0,-2])
```

```python
# линейная комбинация векторов v и w с коэффициентами 2 и -3
d=2*v+-3*w
print('linear combination:', d)
```

```python
# ортогональность линейной комбинации и вектора u
print('ортогональность:',np.dot(u,d))
```
